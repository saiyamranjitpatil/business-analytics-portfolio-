```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
```

```python
orders = pd.read_excel('/content/List of Orders.xlsx')
order_details = pd.read_excel('/content/Order Details.xlsx')
sales_target = pd.read_excel('/content/Sales target.xlsx')
```

```python
print("List of Orders")
display(orders.head())

print("\nShape:", orders.shape)

print("\nColumns:")
print(list(orders.columns))

print("\nInfo:")
orders.info()
```

```python
print("Order Details")
display(order_details.head())

print("\nShape:", order_details.shape)

print("\nColumns:")
print(order_details.columns)

print("\nInfo:")
order_details.info()
```

```python
print("Sales Target")
display(sales_target.head())

print("\nShape:", sales_target.shape)

print("\nColumns:")
print(sales_target.columns)

print("\nInfo:")
sales_target.info()
```

```python
print("Missing Values in List of Orders")
display(orders.isnull().sum())

print("\nMissing Values in Order Details")
display(order_details.isnull().sum())

print("\nMissing Values in Sales Target")
display(sales_target.isnull().sum())
```

```python
print("Duplicate Rows in List of Orders:", orders.duplicated().sum())
print("Duplicate Rows in Order Details:", order_details.duplicated().sum())
print("Duplicate Rows in Sales Target:", sales_target.duplicated().sum())
```

```python
orders["Order Date"] = pd.to_datetime(orders["Order Date"])
```

```python
orders.info()
```

```python
merged_data = pd.merge(orders, order_details, on="Order ID")
```

```python
print("Merged Dataset")
display(merged_data.head())

print("\nShape:", merged_data.shape)

print("\nColumns:")
print(merged_data.columns)
```

```python
category_sales = (
    merged_data.groupby("Category")["Amount"]
    .sum()
    .reset_index()
    .sort_values(by="Amount", ascending=False)
)

print("Total Sales by Category")
display(category_sales)
```

```python
order_profit = (
    merged_data.groupby(["Category", "Order ID"])["Profit"]
    .sum()
    .reset_index()
)
```

```python
average_profit = (
    order_profit.groupby("Category")["Profit"]
    .mean()
    .reset_index(name="Average Profit per Order")
)

print("Average Profit per Order by Category")
display(average_profit)
```

```python
profit_margin = (
    merged_data.groupby("Category")[["Profit", "Amount"]]
    .sum()
    .reset_index()
)

profit_margin["Profit Margin (%)"] = (
    profit_margin["Profit"] / profit_margin["Amount"]
) * 100

profit_margin = profit_margin[["Category", "Profit Margin (%)"]]

print("Profit Margin by Category")
display(profit_margin)
```

```python
category_summary = category_sales.merge(average_profit, on="Category")
category_summary = category_summary.merge(profit_margin, on="Category")

print("Category Performance Summary")
display(category_summary)
```

```python
top_sales = category_summary.loc[category_summary["Amount"].idxmax()]
top_profit = category_summary.loc[category_summary["Average Profit per Order"].idxmax()]
top_margin = category_summary.loc[category_summary["Profit Margin (%)"].idxmax()]

print("Highest Sales Category")
display(top_sales)

print("Highest Average Profit per Order")
display(top_profit)

print("Highest Profit Margin")
display(top_margin)
```

```python
low_sales = category_summary.loc[category_summary["Amount"].idxmin()]
low_profit = category_summary.loc[category_summary["Average Profit per Order"].idxmin()]
low_margin = category_summary.loc[category_summary["Profit Margin (%)"].idxmin()]

print("Lowest Sales Category")
display(low_sales)

print("Lowest Average Profit per Order")
display(low_profit)

print("Lowest Profit Margin")
display(low_margin)
```

```python
category_summary = (
    category_sales
    .merge(average_profit, on="Category")
    .merge(profit_margin, on="Category")
)

print("Category Performance Summary")
display(category_summary)
```

## Part 2: Target Achievement Analysis

```python
print("Sales Target Dataset")
display(sales_target.head())

print("\nShape:", sales_target.shape)

print("\nColumns:")
print(list(sales_target.columns))
```

```python
furniture_target = sales_target[sales_target["Category"] == "Furniture"].copy()

print("Furniture Sales Target")
display(furniture_target)
```

```python
print(furniture_target.dtypes)
```

```python
furniture_target["Target Change (%)"] = (
    furniture_target["Target"].pct_change() * 100
)

print("Furniture Target Analysis")
display(furniture_target)
```

```python
furniture_target["Target Change (%)"] = (
    furniture_target["Target Change (%)"].round(2)
)

display(furniture_target)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    furniture_target["Month of Order Date"],
    furniture_target["Target"],
    marker="o"
)

plt.title("Furniture Sales Target Trend")
plt.xlabel("Month")
plt.ylabel("Target")
plt.xticks(rotation=45)

plt.show()
```

```python
highest_change = furniture_target.loc[
    furniture_target["Target Change (%)"].idxmax()
]

lowest_change = furniture_target.loc[
    furniture_target["Target Change (%)"].idxmin()
]

print("Highest Monthly Increase")
display(highest_change)

print("Lowest Monthly Increase")
display(lowest_change)
```

Target Achievement Insights

- Sales targets for furniture have been steadily increasing throughout the year.


- The monthly increases in targets are mostly small and gradual, showing a steady approach to growth.


- The highest month-over-month increase was observed in July 2018 (1.89%), while the smallest increase was observed in February 2019 (0.87%), indicating that the target growth remained stable throughout the year.

- There have been no big drops in targets, which shows confidence in the furniture category.


- Keeping track of real sales compared to the set targets can help make better plans and use resources more effectively.

# Question 1 - Part 3: Regional Performance Insights

```python
state_orders = (
    orders.groupby("State")["Order ID"]
    .nunique()
    .reset_index(name="Order Count")
    .sort_values(by="Order Count", ascending=False)
)

top_states = state_orders.head(5)

print("Top 5 States by Order Count")
display(top_states)
```

```python
top_state_data = merged_data[
    merged_data["State"].isin(top_states["State"])
]

regional_summary = (
    top_state_data.groupby("State")
    .agg(
        Total_Sales=("Amount", "sum"),
        Average_Profit=("Profit", "mean")
    )
    .reset_index()
)

regional_summary = regional_summary.merge(top_states, on="State")

regional_summary = regional_summary[
    ["State", "Order Count", "Total_Sales", "Average_Profit"]
]

regional_summary = regional_summary.sort_values(
    by="Order Count",
    ascending=False
)

print("Regional Performance Summary")
display(regional_summary)
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    regional_summary["State"],
    regional_summary["Total_Sales"]
)

plt.title("Total Sales of Top 5 States")
plt.xlabel("State")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    regional_summary["State"],
    regional_summary["Average_Profit"]
)

plt.title("Average Profit of Top 5 States")
plt.xlabel("State")
plt.ylabel("Average Profit")

plt.show()
```

### Regional Performance Insights

- Madhya Pradesh had the most orders (101) and the highest total sales (105,140), showing a strong and steady group of customers.


- Maharashtra came in second with 90 orders and a good average profit (21.30), which means it is more profitable than many other states with high order numbers.


- Rajasthan made sales similar to Gujarat even though it had more orders, which suggests that the average amount spent per order is not very high.


- Gujarat had positive sales, but the average profit was much lower (5.34), meaning that even though sales are growing, the profits are not keeping up.


- Punjab had the worst performance among the top five states with a negative average profit (-10.15).
This could be due to higher costs, too much discounting, or selling products with lower profit margins.

### Recommendations

- Keep improving business operations in Madhya Pradesh and Maharashtra since they are key areas for both sales and profits.


- Look at pricing, promotions, and products in Gujarat to boost profit levels.


- Do a deep analysis of Punjab’s profitability to find out why there are losses and take steps to fix the issues.


- Focus on cities in Gujarat and Punjab to improve sales, cut costs, and run better marketing efforts to boost performance.
